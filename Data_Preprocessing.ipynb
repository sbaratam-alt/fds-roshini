```python
# ==========================================================
# TITANIC DATASET - DATA PREPROCESSING
# File: Titanic-Dataset-selected-columns.csv
# ==========================================================

# -----------------------------
# Import Required Libraries
# -----------------------------

import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler

# -----------------------------
# Read Dataset
# -----------------------------

df = pd.read_csv("Titanic-Dataset-selected-columns.csv")

print("="*60)
print("ORIGINAL DATASET")
print("="*60)

print(df.head())

# -----------------------------
# Dataset Information
# -----------------------------

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

# ==========================================================
# 1. Check Missing Values
# ==========================================================

print("\nMissing Values")
print(df.isnull().sum())

# ==========================================================
# 2. Handle Missing Values
# ==========================================================

# Fill Age with Mean

if 'Age' in df.columns:
    df['Age'].fillna(df['Age'].mean(), inplace=True)

# Fill Embarked with Mode

if 'Embarked' in df.columns:
    df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Fill Fare with Median

if 'Fare' in df.columns:
    df['Fare'].fillna(df['Fare'].median(), inplace=True)

print("\nMissing Values After Filling")
print(df.isnull().sum())

# ==========================================================
# 3. Remove Duplicate Records
# ==========================================================

print("\nDuplicate Rows Before:")
print(df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicate Rows After:")
print(df.duplicated().sum())

# ==========================================================
# 4. Encode Categorical Variables
# ==========================================================

le = LabelEncoder()

categorical_columns = ['Sex', 'Embarked']

for col in categorical_columns:
    if col in df.columns:
        df[col] = le.fit_transform(df[col].astype(str))

print("\nEncoded Dataset")
print(df.head())

# ==========================================================
# 5. Feature Scaling
# ==========================================================

# -----------------------------
# Min-Max Normalization
# -----------------------------

scaler = MinMaxScaler()

scale_cols = ['Age', 'Fare']

for col in scale_cols:
    if col in df.columns:
        df[col + '_Normalized'] = scaler.fit_transform(df[[col]])

# -----------------------------
# Standardization
# -----------------------------

std_scaler = StandardScaler()

for col in scale_cols:
    if col in df.columns:
        df[col + '_Standardized'] = std_scaler.fit_transform(df[[col]])

print("\nScaled Data")
print(df.head())

# ==========================================================
# 6. Outlier Detection Using IQR
# ==========================================================

if 'Fare' in df.columns:

    Q1 = df['Fare'].quantile(0.25)
    Q3 = df['Fare'].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df['Fare'] < lower) | (df['Fare'] > upper)]

    print("\nNumber of Outliers in Fare:")
    print(len(outliers))

# ==========================================================
# 7. Separate Features and Target
# ==========================================================

if 'Survived' in df.columns:

    X = df.drop('Survived', axis=1)
    y = df['Survived']

    print("\nFeature Matrix Shape:")
    print(X.shape)

    print("Target Variable Shape:")
    print(y.shape)

# ==========================================================
# Save Preprocessed Dataset
# ==========================================================

df.to_csv("Titanic_Preprocessed.csv", index=False)

print("\nPreprocessed dataset saved successfully.")
```
