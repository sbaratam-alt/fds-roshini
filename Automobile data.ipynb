```python
# ==========================================================
# AUTOMOBILE DATASET
# Plot Various Visualization Plots
# File Name: Automobile_data (1).csv
# ==========================================================

# Import Required Libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set(style="whitegrid")

# ----------------------------------------------------------
# Read Dataset
# ----------------------------------------------------------

df = pd.read_csv("Automobile_data (1).csv")

# ----------------------------------------------------------
# Display Dataset Information
# ----------------------------------------------------------

print("="*60)
print("AUTOMOBILE DATASET")
print("="*60)

print("\nFirst Five Rows")
print(df.head())

print("\nLast Five Rows")
print(df.tail())

print("\nShape of Dataset")
print(df.shape)

print("\nColumn Names")
print(df.columns)

print("\nDataset Information")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nStatistical Summary")
print(df.describe())

# ==========================================================
# 1. Histogram of Numerical Attributes
# ==========================================================

df.hist(figsize=(15,10))
plt.suptitle("Histogram of Numerical Attributes")
plt.show()

# ==========================================================
# 2. Correlation Heatmap
# ==========================================================

plt.figure(figsize=(12,8))
corr = df.corr(numeric_only=True)

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            linewidths=0.5)

plt.title("Correlation Heatmap")
plt.show()

# ==========================================================
# 3. Box Plot (Price)
# ==========================================================

if "price" in df.columns:

    plt.figure(figsize=(6,5))

    sns.boxplot(y=df["price"])

    plt.title("Box Plot of Price")

    plt.show()

# ==========================================================
# 4. Distribution Plot (Price)
# ==========================================================

if "price" in df.columns:

    plt.figure(figsize=(7,5))

    sns.histplot(df["price"],
                 bins=20,
                 kde=True)

    plt.title("Distribution of Price")

    plt.xlabel("Price")

    plt.show()

# ==========================================================
# 5. Count Plot (Fuel Type)
# ==========================================================

if "fuel-type" in df.columns:

    plt.figure(figsize=(6,5))

    sns.countplot(x="fuel-type",
                  data=df)

    plt.title("Fuel Type Count")

    plt.show()

# ==========================================================
# 6. Count Plot (Body Style)
# ==========================================================

if "body-style" in df.columns:

    plt.figure(figsize=(8,5))

    sns.countplot(x="body-style",
                  data=df)

    plt.title("Body Style Count")

    plt.xticks(rotation=30)

    plt.show()

# ==========================================================
# 7. Scatter Plot (Horsepower vs Price)
# ==========================================================

if "horsepower" in df.columns and "price" in df.columns:

    plt.figure(figsize=(7,5))

    sns.scatterplot(x="horsepower",
                    y="price",
                    data=df)

    plt.title("Horsepower vs Price")

    plt.show()

# ==========================================================
# 8. Scatter Plot (Engine Size vs Price)
# ==========================================================

if "engine-size" in df.columns and "price" in df.columns:

    plt.figure(figsize=(7,5))

    sns.scatterplot(x="engine-size",
                    y="price",
                    data=df)

    plt.title("Engine Size vs Price")

    plt.show()

# ==========================================================
# 9. Line Plot (Highway MPG vs Price)
# ==========================================================

if "highway-mpg" in df.columns and "price" in df.columns:

    plt.figure(figsize=(7,5))

    sns.lineplot(x="highway-mpg",
                 y="price",
                 data=df)

    plt.title("Highway MPG vs Price")

    plt.show()

# ==========================================================
# 10. Pair Plot
# ==========================================================

columns = []

for col in ["horsepower", "engine-size", "price", "city-mpg"]:

    if col in df.columns:
        columns.append(col)

if len(columns) >= 2:

    sns.pairplot(df[columns])

    plt.show()

# ==========================================================
# 11. Bar Plot (Average Price by Fuel Type)
# ==========================================================

if "fuel-type" in df.columns and "price" in df.columns:

    plt.figure(figsize=(6,5))

    sns.barplot(x="fuel-type",
                y="price",
                data=df)

    plt.title("Average Price by Fuel Type")

    plt.show()

# ==========================================================
# 12. Pie Chart (Fuel Type Distribution)
# ==========================================================

if "fuel-type" in df.columns:

    plt.figure(figsize=(6,6))

    df["fuel-type"].value_counts().plot(
        kind="pie",
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title("Fuel Type Distribution")

    plt.ylabel("")

    plt.show()

# ==========================================================
# END OF PROGRAM
# ==========================================================

print("\nAll visualization plots generated successfully.")
```
